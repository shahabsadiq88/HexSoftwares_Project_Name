```python
from groq import Groq, BadRequestError
import os, json, re
import numpy as np
from sentence_transformers import SentenceTransformer
from rank_bm25 import BM25Okapi
import faiss
from google.colab import userdata

os.environ["GROQ_API_KEY"] = userdata.get('GROQ_API_KEY')
client = Groq()

# ============================================
# STEP 1: FAQ knowledge base
# ============================================
faq_documents = [
    "TechMart offers free shipping on orders over 5000 rupees within Lahore.",
    "Returns are accepted within 14 days of purchase with original receipt.",
    "We accept payment via JazzCash, EasyPaisa, credit card, and cash on delivery.",
    "TechMart warranty covers manufacturing defects for 1 year on all laptops.",
    "Store hours are 10 AM to 9 PM, Monday through Saturday.",
    "Order tracking is available via SMS after dispatch.",
    "Exchanges are allowed within 7 days if the product is unused and in original packaging.",
    "TechMart offers 0% markup installment plans for 3, 6, and 12 months on orders above 20000 rupees.",
    "Our main store is located at Liberty Market, Lahore, with a second branch in DHA Phase 5.",
    "For support, customers can contact us via WhatsApp at 0300-1234567 or email support@techmart.pk."
]

# ============================================
# STEP 2: Embeddings + BM25 setup
# ============================================
embed_model = SentenceTransformer('all-MiniLM-L6-v2')
doc_embeddings = embed_model.encode(faq_documents)
dimension = doc_embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)
index.add(np.array(doc_embeddings).astype('float32'))

tokenized_docs = [doc.lower().split() for doc in faq_documents]
bm25 = BM25Okapi(tokenized_docs)

def hybrid_search(query, top_k=2):
    query_embedding = embed_model.encode([query]).astype('float32')
    distances, indices = index.search(query_embedding, len(faq_documents))
    semantic_dict = {faq_documents[i]: 1 - distances[0][j] for j, i in enumerate(indices[0])}
    tokenized_query = query.lower().split()
    bm25_scores = bm25.get_scores(tokenized_query)
    keyword_dict = {faq_documents[i]: bm25_scores[i] for i in range(len(faq_documents))}
    sem_values = np.array([semantic_dict[doc] for doc in faq_documents])
    key_values = np.array([keyword_dict[doc] for doc in faq_documents])
    norm_semantic = (sem_values - sem_values.min()) / (sem_values.max() - sem_values.min() + 1e-9)
    norm_keyword = (key_values - key_values.min()) / (key_values.max() - key_values.min() + 1e-9)
    final_score = 0.5 * norm_semantic + 0.5 * norm_keyword
    sorted_indices = np.argsort(final_score)[::-1][:top_k]
    return [faq_documents[i] for i in sorted_indices]

# ============================================
# STEP 3: Tools & Manual Parsing
# ============================================
tools = [{"type": "function", "function": {"name": "search_faq", "description": "Search TechMart FAQ", "parameters": {"type": "object", "properties": {"query": {"type": "string"}}, "required": ["query"]}}}]
def search_faq(query):
    return " ".join(hybrid_search(query))

def run_faq_bot(user_query):
    messages = [{"role": "system", "content": "You are a support bot for TechMart. Search the FAQ for answers. If you need to search, just call search_faq."}, {"role": "user", "content": user_query}]

    try:
        response = client.chat.completions.create(model="llama-3.3-70b-versatile", messages=messages, tools=tools, tool_choice="auto", temperature=0)
        msg = response.choices[0].message
        if msg.tool_calls:
            tool_call = msg.tool_calls[0]
            result = search_faq(json.loads(tool_call.function.arguments)['query'])
            messages.append(msg)
            messages.append({"role": "tool", "tool_call_id": tool_call.id, "name": "search_faq", "content": result})
            final = client.chat.completions.create(model="llama-3.3-70b-versatile", messages=messages, temperature=0)
            return final.choices[0].message.content
        return msg.content

    except BadRequestError as e:
        err_msg = str(e)
        if "failed_generation" in err_msg:
            match = re.search(r'"query":\s*"([^"]+)"', err_msg)
            if match:
                query_extracted = match.group(1)
                result = search_faq(query_extracted)
                messages.append({"role": "assistant", "content": f"Searching for: {query_extracted}"})
                messages.append({"role": "user", "content": f"Based on these FAQ results: {result}, please answer my original question: {user_query}"})
                final = client.chat.completions.create(model="llama-3.3-70b-versatile", messages=messages, temperature=0)
                return final.choices[0].message.content
        return f"Sorry, I encountered an error: {e}"

# Run extended tests
test_queries = [
    "Do you offer free shipping?",
    "What payment methods do you accept?",
    "Where are your stores located?",
    "What are your store timings?",
    "Is there a warranty on laptops?",
    "How can I track my order?",
    "What is your return window?",
    "Do you have installment plans?"
]
for q in test_queries:
    print(f"Q: {q}\nA: {run_faq_bot(q)}\n---")
```
